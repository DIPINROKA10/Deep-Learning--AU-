# Project 10: Overfitting and Underfitting Analysis

**Dataset:** Iris (150 flowers, 4 features, 3 classes)

**Goal:** Build 3 neural networks of *different size* and train them on the **same**
train/test data. Then compare training accuracy vs testing accuracy to find:

| Situation | Meaning |
|---|---|
| Training accuracy **low**, testing accuracy **low** | **Underfitting** (model too simple) |
| Training accuracy **high**, testing accuracy **close to it** | **Good fit** (model just right) |
| Training accuracy **very high**, testing accuracy **much lower** | **Overfitting** (model too complex) |

**Tasks**
1. Build simple neural networks with different numbers of neurons (Iris dataset).
2. Train and evaluate every model on the same training and testing data.
3. Identify underfitting / overfitting and suggest how to improve them.

## Task 1 - Load the data

We split the data **once** and reuse the *exact same* split for all 3 models,
so the only thing that changes between models is the network itself.

```python
# works in Jupyter and in Google Colab
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# make results reproducible
np.random.seed(42)
tf.random.set_seed(42)

print("TensorFlow:", tf.__version__)
```

```python
iris = load_iris()
X, y = iris.data, iris.target

print("Features shape :", X.shape)
print("Labels shape   :", y.shape)
print("Class names    :", list(iris.target_names))

# ONE split, used by every model (70% train, 30% test)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y)

# scale the features (fit only on training data)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("\nTraining samples :", X_train.shape[0])
print("Testing samples  :", X_test.shape[0])
```

## Task 2 - Build 3 models of different complexity

| Model | Architecture | What we expect |
|---|---|---|
| **A** | no hidden layer (`4 -> 3`) | too simple -> **underfitting** |
| **B** | 1 hidden layer, 16 neurons | **good fit** |
| **C** | 3 hidden layers x 64 neurons | too complex -> **overfitting** |

```python
def train_and_evaluate(name, model, epochs=300):
    """Compile, train and score one model. Returns the numbers + history."""
    model.compile(optimizer="adam",
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])

    history = model.fit(X_train, y_train,
                        validation_data=(X_test, y_test),
                        epochs=epochs,
                        verbose=0)

    train_loss, train_acc = model.evaluate(X_train, y_train, verbose=0)
    test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)

    return {"name": name,
            "train_acc": train_acc, "test_acc": test_acc,
            "train_loss": train_loss, "test_loss": test_loss,
            "history": history}


results = []
```

### Model A - too simple (no hidden layer at all)

The network is only a single layer, so it can only draw a **straight** decision
boundary. It is trained for only a few epochs, so it never even learns properly.

```python
model_A = keras.Sequential([
    keras.layers.Input(shape=(4,)),
    keras.layers.Dense(3, activation="sigmoid"),   # no hidden layer
])

model_A.summary()
results.append(train_and_evaluate("A: too simple", model_A, epochs=15))
```

### Model B - one hidden layer with 16 neurons (right size)

```python
model_B = keras.Sequential([
    keras.layers.Input(shape=(4,)),
    keras.layers.Dense(16, activation="relu"),     # 1 hidden layer
    keras.layers.Dense(3, activation="softmax"),
])

model_B.summary()
results.append(train_and_evaluate("B: right size", model_B, epochs=300))
```

### Model C - three hidden layers with 64 neurons each (too complex)

Many more parameters than training samples -> it can **memorise** the training
data instead of learning the general rule. It is trained for many epochs.

```python
model_C = keras.Sequential([
    keras.layers.Input(shape=(4,)),
    keras.layers.Dense(64, activation="relu"),     # hidden layer 1
    keras.layers.Dense(64, activation="relu"),     # hidden layer 2
    keras.layers.Dense(64, activation="relu"),     # hidden layer 3
    keras.layers.Dense(3, activation="softmax"),
])

model_C.summary()
results.append(train_and_evaluate("C: too complex", model_C, epochs=500))
```

### Comparison of the 3 models

`Gap = training accuracy - testing accuracy`. A **big gap** is the sign of overfitting.

```python
print(f"{'Model':18} {'Train Acc':>10} {'Test Acc':>10} {'Gap':>8}")
print("-" * 50)
for r in results:
    gap = r["train_acc"] - r["test_acc"]
    print(f"{r['name']:18} {r['train_acc']:10.3f} {r['test_acc']:10.3f} {gap:8.3f}")
```

```python
# bar chart: train vs test accuracy for each model
x = np.arange(len(results))
width = 0.35

plt.figure(figsize=(7, 4))
plt.bar(x - width / 2, [r["train_acc"] for r in results], width, label="Train accuracy")
plt.bar(x + width / 2, [r["test_acc"] for r in results], width, label="Test accuracy")
plt.xticks(x, [r["name"] for r in results])
plt.ylim(0.4, 1.05)
plt.ylabel("Accuracy")
plt.title("Train vs Test accuracy for the 3 models")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# learning curves: training loss (solid) vs testing loss (dashed)
fig, axes = plt.subplots(1, 3, figsize=(14, 3.5))
for ax, r in zip(axes, results):
    h = r["history"].history
    ax.plot(h["loss"], label="train loss")
    ax.plot(h["val_loss"], "--", label="test loss")
    ax.set_title(r["name"])
    ax.set_xlabel("epoch")
    ax.legend()
    ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Task 3 - Identify the problem of each model

Read the table printed above. The rule of thumb is:

| Train acc | Test acc | Diagnosis |
|---|---|---|
| low | low (similar) | **Underfitting** - the model is too simple to learn the data |
| high | close to train acc | **Good fit** - the model learned the general rule |
| very high | clearly lower | **Overfitting** - the model memorised the training data |

**Model A - Underfitting**
* Both training and testing accuracy are low, and the gap between them is small.
* The model is too simple (no hidden layer, few epochs), so it cannot capture the
  decision boundaries of the Iris data.

**Model B - Good fit**
* Training and testing accuracy are both high and almost equal.
* The model learned the general pattern, not the noise.

**Model C - Overfitting**
* Training accuracy reaches ~100% (it memorised everything) while testing accuracy
  is clearly lower, so the gap is large.
* In the learning curves you can see the training loss keep falling while the
  testing loss starts rising again - the classic sign of overfitting.

### How to improve the models

**To fix underfitting (Model A):**
1. Train for more epochs.
2. Add hidden layers / more neurons so the model has more capacity.
3. Use a non-linear activation (ReLU) instead of only a linear boundary.

**To fix overfitting (Model C):**
1. Add **Dropout** layers.
2. Add **L2 regularization** on the weights.
3. Use **early stopping** - stop when the validation loss stops improving.
4. Reduce the number of neurons / layers (a smaller model).
5. Get **more training data** (or data augmentation).
6. Use k-fold cross-validation to check the model on different splits.

```python
model_fixed = keras.Sequential([
    keras.layers.Input(shape=(4,)),
    keras.layers.Dense(64, activation="relu"),
    keras.layers.Dropout(0.3),            # <== regularisation
    keras.layers.Dense(64, activation="relu"),
    keras.layers.Dropout(0.3),            # <== regularisation
    keras.layers.Dense(3, activation="softmax"),
])

model_fixed.compile(optimizer="adam",
                    loss="sparse_categorical_crossentropy",
                    metrics=["accuracy"])

early_stop = keras.callbacks.EarlyStopping(
    monitor="val_loss", patience=20, restore_best_weights=True)

history_fixed = model_fixed.fit(X_train, y_train,
                                validation_data=(X_test, y_test),
                                epochs=500, callbacks=[early_stop], verbose=0)

_, tr = model_fixed.evaluate(X_train, y_train, verbose=0)
_, te = model_fixed.evaluate(X_test, y_test, verbose=0)
print(f"Model C fixed : train acc = {tr:.3f} | test acc = {te:.3f} | gap = {tr - te:.3f}")
```
