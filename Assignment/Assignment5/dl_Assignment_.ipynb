Activation Function Using Python

```python
import numpy as np
import matplotlib.pyplot as plt

#Task 1 : define activation function

#sigmoid functions
def sigmoid(x):
  return 1/(1+ np.exp(-x))


#relu function
def relu(x):
  return np.maximum(0,x)


# Tanh Function
def tanh(x):
  return np.tanh(x)
```

```python
#Task 2



#input values
x_value = np.array([-5,-3,-1,0,1,3,5])


#Apply activation finctions
sigmoid_output = sigmoid(x_value)
relu_output = relu(x_value)
tanh_output = tanh(x_value)


#display outputs
print("input Values")
print(x_value)

print("\nSigmoid Output")
print(sigmoid_output)
```

```python
#Task 3
#generated continous input values for plotting

x= np.linspace(-5,5,100)


#calculate outputs
y_sigmoid = sigmoid(x)
y_relu = relu(x)
y_tanh = tanh(x)

#plot all  activation functions
plt.figure(figsize=(10,6))
plt.plot(x,y_sigmoid,label='sigmoid')
plt.plot(x,y_relu,label='relu')
plt.plot(x,y_tanh,label='tanh')


plt.title("Comparison Of Activation Finctions")
plt.xlabel("input value")
plt.ylabel("output Value")

plt.grid(True)
plt.legend()
plt.show()
```
